#### **Función auxiliar: Factorial**
Esta función es la base de ambos problemas. Calcula `n!` de forma
iterativa y es reutilizada por todas las funciones posteriores.

```python
def factorial(n):
    # casos especiales: n es 0 o negativo
    if n == 0 :
        return 1
    elif n < 0 :
        return ("Error: el factorial no está definido para números negativos")
    # caso general: n!
    else :
      fact = 1
      for i in range (1,n+1):
        fact = fact * i
      return fact
```

## **Problema 2: Calculadora general de combinaciones**


#### **1. Descripción**
Usamos una combinación para contar el número de formas de escoger `r` objetos
de un conjunto de `n` objetos distintos, sin importar el orden. Este programa calcula la combinación de cualquier entrada que digite el usuario.



#### **2. Fórmula**
$$C(n, r) = \binom{n}{r} = \frac{n!}{r!(n-r)!}$$

La división por `r!` elimina los arreglos repetidos que surgen
de ordenar de distintas formas los mismos `r` objetos escogidos.



#### **3. Algoritmo**
**3.1. Combinaciones:**
  - Validar que `r >= 0` y `r <= n`.
  - Calcular `n!` de forma iterativa.
  - Calcular `r!` y `(n-r)!` de la misma forma.
  - Retornar la división entera `n! // (r! * (n-r)!)`.

Se usa división entera `//` para evitar imprecisiones de punto flotante,
garantizando siempre un resultado entero exacto.

**3.2. Verificar si cumple la identidad:** $$\binom{n}{r} = \binom{n}{n-r}$$
  - Calcular `C(n,r)` y `C(n,n-r)` usando la función `combinaciones`.
  - Comparar ambos valores y retornar un mensaje indicando si son iguales.

**3.3. Fila de Pascal:**
  - Para cada `i` en el rango de 0 a n, calcular C(n,i) usando la función `combinaciones`.
  - Agregar cada resultado a una lista.
  - Retornar la lista completa.

**3.4. Triangulo de pascal:**
  - Encontrar el número más grande (el centro de la última fila).
  - Calcular el ancho de la celda, el cual será la cantidad de dígitos del número máximo + 2 espacios de margen.
  - Aseguramos que el ancho sea un número par para que la división sea exacta.
  - Dibujar el triángulo con el ancho automatizado

#### **4. Codigo**

```python
def combinaciones(n,r):
   # validaciones: qué condiciones deben cumplir n y r
   if r > n:
      return ("Error: no puede escoger una cantidad mayor a los elementos que hay")
   elif r < 0:
      return ("Error: no puede escoger un número negativo de elementos")
   # cálculo: la fórmula
   else :
      return factorial(n)//(factorial(r) * factorial(n-r))
```

```python
def verificar_identidad(n,r):
  c_nr = combinaciones(n,r)
  c_nnr = combinaciones(n, n-r)
  if c_nr == c_nnr :
    return (f" - Cumple la identidad, pues C({n},{r}) = {c_nr} == C({n},{n-r}) = {c_nnr} ")
  else:
    return (f" - No cumple la identidad, pues C({n},{r}) = {c_nr} != C({n},{n-r}) = {c_nnr} ")
```

```python
def fila_pascal(n):
  fila = []
  for i in range (0, n+1):
    fila.append (combinaciones(n, i))
  return fila

def triangulo_de_pascal(n):
    # 1. Encontramos el número más grande (el centro de la última fila)
    ultima_fila = fila_pascal(n)
    numero_maximo = max(ultima_fila)

    # 2. El ancho ideal será la cantidad de dígitos del número máximo + 2 espacios de margen
    # Esto asegura que los números nunca se peguen entre sí.
    ancho_celda = len(str(numero_maximo)) + 2

    # 3. Nos aseguramos de que el ancho sea un número par para que la división sea exacta
    if ancho_celda % 2 != 0:
        ancho_celda += 1

    # 4. Dibujamos el triángulo con el ancho automatizado
    for i in range(0, n+1):
        numeros = fila_pascal(i)
        fila_formateada = "".join(str(x).center(ancho_celda) for x in numeros)
        margen_izquierdo = " " * ((n - i) * (ancho_celda // 2))
        print(margen_izquierdo + fila_formateada)
```

## **Problema 9: Coeficientes multinomiales y palabras con letras repetidas**

#### **1. Descripción**
El coeficiente multinomial cuenta cuántas palabras distintas pueden formarse
con las letras de una palabra que tiene letras repetidas. Por ejemplo, con
las letras de "BANANA" no se obtienen 6! = 720 palabras distintas, sino menos,
porque intercambiar las A's entre sí no produce una palabra nueva.



#### **2. Fórmula**
$$\frac{n!}{n_1! \cdot n_2! \cdots n_k!}$$

Donde `n` es la longitud total de la palabra y `n1, n2, ..., nk` son las
frecuencias de cada letra distinta. Se divide por cada $$n_i$$ para eliminar
los arreglos repetidos causados por las letras iguales.



#### **3. Algoritmo**
**3.1. Coeficiente desde una palabra:**
   - Usar la función `Counter` para contar la frecuencia de cada letra.
   - Calcular `n!` donde `n` es la longitud de la palabra.
   - Calcular el factorial de cada frecuencia y multiplicarlos entre sí.
   - Retornar la división entera `n! // (n1! * n2! * ... * nk!)`.

**3.2. Coeficiente desde una lista de cantidades:**
   - Sumar todos los valores de la lista para obtener `n`.
   - Calcular el factorial de cada valor y multiplicarlos entre sí.
   - Retornar la división entera `n! // (n1! * n2! * ... * nk!)`.

**3.3. Listar palabras distintas (para palabras de longitud ≤ 8):**
   - Caso base: si la palabra tiene 1 letra, retornar ella misma.
   - Para cada letra, construir la palabra sin esa letra (el resto).
   - Llamar recursivamente a la función con el resto.
   - Agregar cada combinación de letra + permutación del resto a un conjunto.
   - El conjunto elimina automáticamente los duplicados.

####**4. Codigo**

```python
from collections import Counter

def coeficiente_multinomial(palabra):
  frecuencias = Counter(palabra)
  factorial_palabra = factorial(len(palabra))
  denominador = 1

  for freq in frecuencias.values():
      denominador *= factorial(freq)

  coeficiente = factorial_palabra // denominador

  return (f"El coeficiente multinomial de {palabra} es {coeficiente}")
```

```python
def coeficiente_por_cantidades(lista):
  n = sum(lista)
  denominador = 1
  for freq in lista:
      denominador *= factorial(freq)
  coeficiente = factorial(n) // denominador
  return (f"El coeficiente multinomial de la lista es {coeficiente}")
```

```python
def listar_palabras(palabra):
    if len(palabra) > 8:
        return "Palabra muy larga para listar"

    if len(palabra) == 1:
        return {palabra}

    resultado = set()
    for i, letra in enumerate(palabra):
        resto = palabra[:i] + palabra[i+1:]  # palabra sin la letra i
        for perm in listar_palabras(resto):
            resultado.add(letra + perm)

    return resultado
```

####**5. Pruebas**

```python
print(coeficiente_multinomial("BANANA"))       # 60
print(coeficiente_multinomial("MISSISSIPPI"))  # 34650
print(coeficiente_multinomial("ANA"))          # 3
print(coeficiente_por_cantidades([4, 3, 2]))   # 1260
print(coeficiente_por_cantidades([2, 2, 2]))   # 90
```

####**6. Casos especiales**

```python
print(coeficiente_multinomial("A"))        # 1 — una sola letra
print(coeficiente_multinomial("AA"))       # 1 — todas iguales
print(coeficiente_multinomial("AB"))       # 2 — todas distintas
print(coeficiente_por_cantidades([1]))     # 1 — un solo grupo
print("Palabras distintas de ANA:", listar_palabras("ANA"))             # {'ANA', 'NAA', 'AAN'}
```

#### **7. Eficiencia**
- `coeficiente_multinomial` es **O(n)**: recorre las frecuencias una vez.
- `coeficiente_por_cantidades` es **O(n)**: recorre la lista una vez.
- `listar_palabras` es **O(n!)** en el peor caso: genera todas las
  permutaciones posibles. Por eso se limita a palabras de longitud ≤ 8,
  pues 8! = 40320 ya es un número considerable de combinaciones.
